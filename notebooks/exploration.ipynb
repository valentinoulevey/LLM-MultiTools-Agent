## 4. Agent complet (nécessite OPENROUTER_API_KEY dans .env)

```python
from agent.graph import build_agent
from memory.checkpointer import get_checkpointer
from langchain_core.messages import HumanMessage

agent = build_agent(checkpointer=get_checkpointer())
config = {'configurable': {'thread_id': 'demo'}}
res = agent.invoke({'messages': [HumanMessage(content='Quel temps fera-t-il demain a Lyon ?')], 'tools_used': []}, config=config)
print('Outils:', res['tools_used'])
print(res['final_answer'])
```
